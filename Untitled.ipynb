```python
# backend/accounts/views_update.py
import time
import traceback
from pathlib import Path
from datetime import timedelta

import ccxt, time
import numpy as np
import pandas as pd
from django.http import JsonResponse, HttpResponseBadRequest
from django.views.decorators.csrf import csrf_exempt

DATA_DIR = Path(__file__).resolve().parent.parent / "data"
DATA_DIR.mkdir(exist_ok=True)

ET = "America/New_York"
STEP_MS = 5 * 60 * 1000  # 5-minute candles
SKIP_MS = 24 * 60 * 60 * 1000  # if empty batch, jump ahead 1 day

# EXACTLY like your Dash script: coinbase + slash symbols
SYMBOL_MAP = {
    "BTC": "BTC/USD",
    "SOL": "SOL/USD",
    "ADA": "ADA/USD",
}

def _to_utc_ms_from_et(ts_et) -> int:
    t = pd.Timestamp(ts_et)
    if t.tzinfo is None:
        t = t.tz_localize(ET, nonexistent="shift_forward")
    return int(t.tz_convert("UTC").timestamp() * 1000)

def _ensure_numeric(df, cols):
    for c in cols:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors="coerce")

@csrf_exempt
def api_get_csv(request):
    print("DEBUG: api_get_csv called; method =", request.method)
    if request.method != "POST":
        return HttpResponseBadRequest("POST only")

    print("DEBUG: request.POST =", request.POST)
    raw_symbol = request.POST.get("symbol")
    print("DEBUG: raw_symbol =", raw_symbol)
    symbol_key = (raw_symbol or "").upper()

    if symbol_key not in SYMBOL_MAP:
        print("DEBUG: unsupported symbol ->", symbol_key)
        return JsonResponse({"ok": False, "error": "Unsupported symbol"}, status=400)

    ccxt_symbol = SYMBOL_MAP[symbol_key]
    csv_path = DATA_DIR / f"{symbol_key.lower()}usd_5m_coinbase.csv"
    print(f"DEBUG: ccxt_symbol={ccxt_symbol}  csv_path={csv_path}")

    try:
        ex = ccxt.coinbase({"enableRateLimit": True})
        ex.load_markets()
        print("DEBUG: coinbase markets loaded (len):", len(ex.symbols))
        # Optional deep check:
        # print("DEBUG: sample symbols:", ex.symbols[:50])

        # Determine start point
        if csv_path.exists():
            df = pd.read_csv(csv_path, low_memory=False)
            if "ts" not in df.columns:
                df["ts"] = pd.to_datetime(df["date"] + " " + df["time"], errors="coerce")
            else:
                df["ts"] = pd.to_datetime(df["ts"], errors="coerce")
            df = df.dropna(subset=["ts"]).sort_values("ts").reset_index(drop=True)
            if df.empty:
                print("DEBUG: existing CSV had no valid rows; treating as fresh history")
                df = pd.DataFrame()
                since_ms = _to_utc_ms_from_et("2015-01-01 00:00:00")
            else:
                last_ts = df["ts"].iloc[-1]
                since_ms = _to_utc_ms_from_et(last_ts + timedelta(minutes=5))
                print("DEBUG: resume from last_ts =", last_ts, "→ since_ms =", since_ms)
        else:
            df = pd.DataFrame()
            # NEW: start ~180 days back so we actually hit data quickly
            end_et = pd.Timestamp.now(tz=ET).tz_localize(None) - timedelta(minutes=5)
            since_ms = _to_utc_ms_from_et(end_et - timedelta(days=365))
            print("DEBUG: fresh history start ~180d back, since_ms =", since_ms)

            print("DEBUG: fresh history start since_ms =", since_ms)

        # End at last completed bar
        end_et = pd.Timestamp.now(tz=ET).tz_localize(None) - timedelta(minutes=5)
        end_ms = _to_utc_ms_from_et(end_et)
        print(f"DEBUG: target window since_ms={since_ms} .. end_ms={end_ms} (end_et={end_et})")

        rows = []
        empty_hits = 0
        while since_ms <= end_ms:
            batch = ex.fetch_ohlcv(ccxt_symbol, timeframe="5m", since=since_ms, limit=300)
            # NEW
            if not batch:
                print("DEBUG: empty batch returned at", since_ms, "— stopping this run")
                break


            rows.extend(batch)
            last_ms = batch[-1][0]
            # normal forward step (exactly like your Dash code)
            new_since = max(last_ms + STEP_MS, since_ms + STEP_MS)
            print(f"DEBUG: batch_len={len(batch)}  last_ms={last_ms}  next since={new_since}")
            since_ms = new_since
            time.sleep(getattr(ex, "rateLimit", 200) / 1000.0)

        if rows:
            print("DEBUG: total fetched rows =", len(rows))
            new = pd.DataFrame(rows, columns=["ms","open","high","low","close","volume"])
            new["ts"] = (
                pd.to_datetime(new["ms"], unit="ms", utc=True)
                  .dt.tz_convert(ET)
                  .dt.tz_localize(None)
            )
            new["date"] = new["ts"].dt.strftime("%Y-%m-%d")
            new["time"] = new["ts"].dt.strftime("%H:%M:%S")
            new = new[["date","time","open","high","low","close","volume","ts"]]
            _ensure_numeric(new, ["open","high","low","close","volume"])

            if df.empty:
                combined = new.copy()
            else:
                combined = (
                    pd.concat([df, new], ignore_index=True)
                      .drop_duplicates(subset=["ts"])
                      .sort_values("ts")
                      .reset_index(drop=True)
                )

            combined.to_csv(csv_path, index=False)
            print(f"DEBUG: wrote CSV → {csv_path} (rows={len(combined)})")
        else:
            print("DEBUG: no rows fetched in this run")
            combined = df  # possibly still empty if first run

        return JsonResponse({
            "ok": True,
            "rows": int(len(combined)),
            "from": (str(combined["ts"].iloc[0]) if not combined.empty else None),
            "to":   (str(combined["ts"].iloc[-1]) if not combined.empty else None),
        })

    except Exception as e:
        traceback.print_exc()
        return JsonResponse({"ok": False, "error": f"{type(e).__name__}: {e}"}, status=500)
```

```python
import ccxt, pandas as pd, time

ex = ccxt.coinbase({"enableRateLimit": True})
ex.load_markets()

SYMBOL = "SOL/USD"
TIMEFRAME = "5m"
TF_MS = ex.parse_timeframe(TIMEFRAME) * 1000
LIMIT = 2023  # ~7 days of 5m candles
ET = "America/New_York"

def to_utc_ms_from_et(ts):
    """Accepts string or Timestamp, naive or tz-aware, and returns UTC ms.
       If naive, assume ET; if tz-aware, convert to UTC."""
    t = pd.Timestamp(ts)
    if t.tzinfo is None:
        t = t.tz_localize(ET, nonexistent="shift_forward")
    return int(t.tz_convert("UTC").timestamp() * 1000)

# Start very early; the exchange will simply return empty until the listing.
since = to_utc_ms_from_et("2015-01-01 00:00:00")

# End at the last fully closed 5-minute bar in ET
now_et = pd.Timestamp.now(tz=ET)
end_et = (now_et.floor("5min") - pd.Timedelta(minutes=5))
end = to_utc_ms_from_et(end_et)

all_rows = []
while since <= end:
    print(since)
    batch = ex.fetch_ohlcv(SYMBOL, timeframe=TIMEFRAME, since=since, limit=LIMIT)
    if not batch:
        # If you hit this at the beginning of history, advance a big step to find first data.
        # Otherwise, it's likely you've reached the true end or a temporary gap.
        # Try jumping forward by ~7 days (LIMIT * TF_MS), otherwise break.
        jump = LIMIT * TF_MS
        probe = since + jump
        if probe <= end:
            since = probe
            continue
        break
    all_rows.extend(batch
    last_ms = batch[-1][0]
    since = max(last_ms + TF_MS, since + TF_MS)  # advance at least one candle
    time.sleep(getattr(ex, "rateLimit", 200) / 1000.0)

df = pd.DataFrame(all_rows, columns=["ms","open","high","low","close","volume"])
if not df.empty:
    df["ts"] = (
        pd.to_datetime(df["ms"], unit="ms", utc=True)
          .dt.tz_convert(ET).dt.tz_localize(None)
    )
    print(df["ts"].min(), "→", df["ts"].max(), "rows:", len(df))
else:
    print("No candles returned (symbol may have listed recently, try a later start).")
```
